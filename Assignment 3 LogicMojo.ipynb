Q3. Feature Scaling using NumPy (Very Important)

Context: Most ML models require normalized data.

Dataset: Iris (NumPy array)

Tasks:

1. Apply standardization:

$$
X' = \frac{X - \mu}{\sigma}
$$
	​2. 

Verify:
Mean ≈ 0
Std ≈
on

```python
columns = list(samples.columns)

samples_scaled = samples.copy(deep = True)

for col in columns:
    print(f"Mean of {col} : ", samples[col].mean())
    print(f"std dev of {col} : ", samples[col].std())

for col in columns:
    samples_scaled[col] = (samples[col] - samples[col].mean())/samples[col].std()

print(samples_scaled)  
```

Std. Dev is 1 and Mean is 0 means the data is normally distributed. We donot need to scale the data.

Q5. Filtering Data for Business Logic

Q7. Feature Engineering for Model Improvement

Context: Raw data is rarely useful—features must be created.

Dataset: Titanic

Tasks:

1. Create:
family_size = sibsp + parch + 1
2. Create:
is_alone (binary feature)
3. Compute survival rate by is_alone
4. Explain:
Why engineered features improve ML models

```python
data['family_size'] = data['sibsp'] + data['parch'] + 1
data['is_alone'] = np.where(data['sibsp'] + data['parch'] == 0, 1, 0)

data.groupby('is_alone')['survived'].mean()
```

adding well engineered features like family size and is_alone can provide additional insights into the survival rates of passengers. For instance, passengers who were alone might have had different survival chances compared to those with family members on board. By analyzing the survival rates based on these new features, we can gain a deeper understanding of the factors that influenced survival during the Titanic disaster.

Q11. End-to-End ML Preprocessing Pipeline (Titanic)

Context: Prepare dataset for ML model training

Tasks:

1. Load dataset
2. Handle missing values
3. Encode categorical variables (sex, embarked)
4. Normalize numerical features
5. Convert to NumPy array

Output:

Final feature matrix ready for model training

```python
df = sns.load_dataset('titanic')
```

```python
df.isnull().sum() 
```

```python
df.describe()
```

```python
#handling missing values
df.drop(columns = 'deck', axis= 1, inplace=True)
df['age'].fillna(df['age'].mean(),inplace= True)
df['embarked'].fillna(df['embarked'].mode(), inplace= True)

#handle encoding 
df['sex'] = df['sex'].map({'male' : 0, 'female' : 1})
df['embarked'] = df['embarked'].map({'C':0, 'Q': 1, 'S': 2})

# normalize columns 
for col in ['age', 'fare']:
    df[col] = (df[col] - df[col].mean())/df[col].std()

df_np = df.values 
df_np
```

Q13. Custom Feature Transformation using apply()

Context: Real ML requires custom logic transformation.

Dataset: Titanic

Tasks:

1. Create age_group:
Child / Adult / Senior
2. Use .apply()
3. Compute survival rate per group
4. Interpret:
Which segment has highest survival likelihood

```python
titanic = sns.load_dataset('titanic')
```

```python
titanic['age_group'] = titanic['age'].apply(lambda x: 'child' if x < 18 else 'adult' if x < 60 else 'senior')
```

```python
survival_rate_by_age_group = titanic.groupby('age_group')['survived'].mean()
survival_rate_by_age_group
```

Incase of rescue operations women and children are given priority. Hence, the survival rate of children is higher

Q14. Error Analysis Simulation (Model Evaluation Thinking)

Tasks:

1. Generate 1000 random error values using NumPy
2. Compute:
Mean
Standard deviation
3. Identify outliers:
Using mean ± 2*std
4. Explain:
What high variance means in model performance
