Write a program to print 1 to 10 using while loop.

```python
i= 0
n= 10
while i <= n:
    print(i)
    i += 1

```

Program to print sum of first 10 numbers.

```python
n= 10 
sum= 0
for i in range(n+1):
    sum += i

print(sum)
```

Alternate Approach:

```python
n =  10
sum =  n*(n+1)//2
print(sum)
```
