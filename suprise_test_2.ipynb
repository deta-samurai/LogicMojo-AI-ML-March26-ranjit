```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
```

1. Simple Linear Regression - Study Hours vs Marks Scored

```python
df = pd.DataFrame({'hours_studied' : [1,2,3,4,5], 'marks_scored' : [35, 45, 50, 65, 70]})
```

```python
X = df[['hours_studied']]
y = df['marks_scored']
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
# when student studies 6 hours
prediction = model.coef_[0] * 6 + model.intercept_
prediction 
```

3. Polynomial Regression

```python
df3 = pd.DataFrame({'x' : [1,2,3,4,5], 'y' : [2, 8, 18, 32, 50]})
```

```python
X = df3[['x']]
y = df3['y']
```

```python
poly = PolynomialFeatures(degree=2,include_bias=False)
X_poly = poly.fit_transform(X)
```

```python
model3 = LinearRegression()
model3.fit(X_poly, y)
```

```python
print("Slope:", model3.coef_)
print("Intercept:", model3.intercept_)
```

```python
x_new = poly.transform([[6]])
prediction = model3.predict(x_new)
prediction
```

6. Z-score

```python
scores = [45, 50, 55, 60, 65, 70, 75]

mean = np.mean(scores)
std = np.std(scores)        # Population standard deviation

z_score = (75 - mean) / std

print("Mean:", mean)
print("Standard Deviation:", std)
print("Z-score:", z_score)

if abs(z_score) > 2:
    print("75 is an outlier.")
else:
    print("75 is NOT an outlier.")
```

8. Binomial Prob

```python
from scipy.stats import binom

n = 8
p = 0.1
k = 2

probability = binom.pmf(k, n, p)

print(probability)
```

```python
from scipy.stats import norm

mean = 70
std = 8

# (a)
prob_between = norm.cdf(78, mean, std) - norm.cdf(62, mean, std)

# (b)
prob_above = 1 - norm.cdf(86, mean, std)

print("a) Percentage:", prob_between * 100)

print("b) Percentage:", prob_above * 100)
```

10. hypothesis testing

```python
from scipy.stats import ttest_rel

df = pd.DataFrame({
    "before": [70, 72, 68, 75, 74],
    "after":  [78, 80, 76, 82, 79]
})
h_0 = 'No improvement'
h_1 = 'Improvement due to training'
t_stat, p_value = ttest_rel(df["after"], df["before"])

print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print(h_1)

else: 
    print(h_0)
```
