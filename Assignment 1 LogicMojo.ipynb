Question 1:- A digital payment platform recorded the following transaction amounts made by a customer during a single day.

transactions = [1200, 3000, 250, 7800, 540, 3000, 9100, 250, 1500, 7800]

Write a Python program that performs the following analysis:
1. Identify and display all transaction values that appear more than once.
2. Detect transactions greater than ₹5000 as these may require fraud monitoring.
3. Calculate the total spending for the day.
4. Create a new list containing only unique transaction values.
5. Sort the unique transaction list in descending order to view the largest transactions first.

1. Identify and display all transaction values that appear more than once.

```python
transactions = [1200, 3000, 250, 7800, 540, 3000, 9100, 250, 1500, 7800]

#using nested for loops to find duplicate transactions and count their occurrences
for i in range(len(transactions)):
    for j in range(i + 1, len(transactions)):
        if transactions[i] == transactions[j]:
            print(f"Transaction amount: {transactions[i]}, Count: {transactions.count(transactions[i])}")

```

```python
transactions = [1200, 3000, 250, 7800, 540, 3000, 9100, 250, 1500, 7800]

#using python dictionary and its methods.
txn_counter = {}
for transaction in transactions:
    txn_counter[transaction] = txn_counter.get(transaction, 0) + 1

for amount, count in txn_counter.items():
    if count > 1:
        print(f"Transaction amount: {amount}, Count: {count}")
    
```

```python
#using python collections. 
from collections import Counter
transactions = [1200, 3000, 250, 7800, 540, 3000, 9100, 250, 1500, 7800]
transaction_counts = Counter(transactions)

for amount, count in transaction_counts.items():
    if transaction_counts[amount] > 1:
        print(f"Transaction amount: {amount}, Count: {count}")
```

2. Detect transactions greater than ₹5000 as these may require fraud monitoring.

```python
transactions = [1200, 3000, 250, 7800, 540, 3000, 9100, 250, 1500, 7800]

threshold = 5000

for transaction in transactions:
    if transaction > threshold:
        print(f"Transaction amount: {transaction} is greater than {threshold}.")
```

3. Calculate the total spending for the day.

```python
transactions = [1200, 3000, 250, 7800, 540, 3000, 9100, 250, 1500, 7800]
#using sum() function
sum(transactions)
```

```python
transactions = [1200, 3000, 250, 7800, 540, 3000, 9100, 250, 1500, 7800]
#using for loop 
sum = 0
for transaction in transactions:
    sum += transaction

print(sum)
```

4. Create a new list containing only unique transaction values.

```python
transactions = [1200, 3000, 250, 7800, 540, 3000, 9100, 250, 1500, 7800]

#using set() 
new_transactions = list(set(transactions))

new_transactions 
```

```python
transactions = [1200, 3000, 250, 7800, 540, 3000, 9100, 250, 1500, 7800]

#using for loop 
unique_transactions = []
for transaction in transactions:
    if transaction not in unique_transactions:
        unique_transactions.append(transaction)

unique_transactions
```

5. Sort the unique transaction list in descending order to view the largest transactions first.

```python
for i in range(len(unique_transactions)):
    for j in range(i + 1, len(unique_transactions)):

        if unique_transactions[i] < unique_transactions[j]:
            unique_transactions[i], unique_transactions[j] = unique_transactions[j], unique_transactions[i]

unique_transactions
```

ages = [25, 42, -1, 34, None, 52, 17, -5, 29, None, 46]

Write a Python program that:
1. Iterates through the dataset and removes all invalid age values.
2. Stores only valid ages in a new list.
3. Calculates the average age of valid customers.
4. Displays the list of customers whose age is greater than 30.

1. Iterates through the dataset and removes all invalid age values.

```python
ages = [25, 42, -1, 34, None, 52, 17, -5, 29, None, 46]

new_list = []

for age in ages:
    if age is not None and age >= 0:
        new_list.append(age)

new_list
```

3. Calculates the average age of valid customers.

```python
sum = 0

for age in new_list:
    sum += age

print(sum/len(new_list))
```

Displays the list of customers whose age is greater than 30.

```python
for age in new_list :
    if age > 30:
        print(f"Age: {age} is greater than 30.")
```

5. Prints how many invalid entries were removed during the cleaning process.

```python
ctr = 0 
for age in ages:
    if age is None or  age < 0:
        ctr += 1

print(ctr)
```

Question 3:- An analyst wants to compare the price of products listed on two e-commerce platforms.

The following data is provided:
products = ["Laptop","Phone","Tablet","Monitor","Headphones"]
amazon_prices = [75000, 45000, 30000, 20000, 5000]
flipkart_prices = [72000, 47000, 31000, 19500, 4800]

Write a Python program that:
1. Combines the product names and their prices into a single data structure representing a table.
2. Compares the price of each product across both platforms.
3. Determines which platform offers the lower price.
4. Calculates the difference between the two prices.
5. Displays the product name, cheaper platform, and price difference.

1. Combines the product names and their prices into a single data structure representing a table.

```python
products = ["Laptop","Phone","Tablet","Monitor","Headphones"]
amazon_prices = [75000, 45000, 30000, 20000, 5000]
flipkart_prices = [72000, 47000, 31000, 19500, 4800]
products_dict ={}

for i in range(len(products)):
    products_dict[products[i]] = {"amazon_price" : amazon_prices[i], "flipkart_price" : flipkart_prices[i]}

print(products_dict)
```

3. Determines which platform offers the lower price.

```python
for product in products_dict:
    amazon_price = products_dict[product]["amazon_price"]
    flipkart_price = products_dict[product]["flipkart_price"]
    
    if amazon_price < flipkart_price:
        print(f"{product} is cheaper on Amazon with price {amazon_price}.")
        print("Difference between the two prices is: ", flipkart_price - amazon_price)
    elif flipkart_price < amazon_price:
        print(f"{product} is cheaper on Flipkart with price {flipkart_price}.")
        print("Difference between the two prices is: ", amazon_price - flipkart_price)
    else:
        print(f"{product} has the same price on both platforms: {amazon_price}.")
```

Question 4:- A web platform records the hours when users log into the system.

login_hours = [1, 3, 5, 7, 2, 6, 8, 4, 9, 10]
Write a Python program that:

1. Separates login hours into even-hour logins and odd-hour logins.
2. Identifies login hours that fall into peak activity time (hours greater than 5).
3. Sorts the login hours in ascending order.
4. Calculates how many logins occurred during peak hours.

```python
login_hours = [1, 3, 5, 7, 2, 6, 8, 4, 9, 10]

even_hours = []
odd_hours = []
peak_hr_ctr = 0
login_hours.sort()
for hour in login_hours:
    if hour % 2 == 0:
        even_hours.append(hour)
    if hour > 5:
        peak_hr_ctr += 1
    else:
        odd_hours.append(hour)
print("login hours sorted", login_hours)
print("Even login hours: ", even_hours)
print("Odd login hours: ", odd_hours)
print("Peak hours (greater than 5): ", peak_hr_ctr)
```
